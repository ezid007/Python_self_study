```python
import traceback

def function_a():
    print("function_a가 function_b를 호출합니다.")
    function_b()
    print("function_a가 종료됩니다.")

def function_b():
    print("function_b가 function_c를 호출합니다.")
    function_c()
    print("function_b가 종료됩니다.")

def function_c():
    print("function_c에서 오류를 발생시킵니다.")
    # 여기서 ZeroDivisionError 발생!
    result = 10 / 0
    return result

# 메인 코드
try:
    function_a()
    
except Exception:
    # 에러가 발생하면 상세 내용을 문자열로 가져와 출력
    error_details = traceback.format_exc()
    print("\n--- 에러 발생! 상세 정보 ---")
    print(error_details)
    print("--------------------------")
    # 이 정보를 파일에 기록하여 에러 로그를 만들 수 있습니다.
    # with open("error_log.txt", "a") as f:
    #     f.write(error_details + "\n")

print("프로그램이 정상적으로 종료되었습니다.")
```

```python
def c_withdraw(balance, amount):
    if amount > balance:
        return None, "잔액 부족" # 오류 메시지를 함께 반환
    return balance - amount, "성공"

def b_buy_item(cash, item_price):
    new_balance, msg = c_withdraw(cash, item_price)
    if new_balance is None:
        # 오류를 다시 상위로 전달... 매우 번거롭다.
        return None, msg 
    return new_balance, "구매 성공"

def a_start_shopping():
    final_result, final_msg = b_buy_item(50000, 100000)
    if final_result is None:
        # 맨 위에서야 겨우 오류를 처리한다.
        print(f"쇼핑 실패! 이유: {final_msg}")

a_start_shopping()
```

```python
def c_withdraw_safe(balance, amount):
    if amount > balance:
        raise ValueError("잔액 부족") # 오류 발생!
    return balance - amount

def b_buy_item_safe(cash, item_price):
    # 이 함수는 오류 처리에 신경 쓸 필요가 없다!
    # 그냥 자기 할 일만 하면 된다.
    return c_withdraw_safe(cash, item_price)

def a_start_shopping_safe():
    try:
        # 여러 단계를 거친 함수의 결과를 바로 받는다.
        b_buy_item_safe(50000, 100000)
    except ValueError as e:
        # C함수에서 발생한 오류가 중간 단계를 모두 건너뛰고
        # 여기에 직접 배송된다!
        print(f"쇼핑 실패! 이유: {e}")

a_start_shopping_safe()
```

```python
# '은행 계좌'의 역할을 하는 클래스
class BankAccount:
    def __init__(self, initial_balance=0):
        # 잔액(_balance)은 이 클래스 외부에서 함부로 접근할 수 없는 중요한 정보.
        self._balance = initial_balance

    def withdraw(self, amount):
        """출금을 담당하는 유일한 창구(메서드)"""
        print(f"[은행 시스템] {amount:,}원 출금 요청 확인...")
        
        # 출금 가능 여부는 '은행 시스템' 스스로 판단한다.
        if self._balance < amount:
            # 판단 결과, 불가능하면 여기서 직접 에러를 발생시킨다!
            raise ValueError("잔액이 부족하여 출금할 수 없습니다.")
        
        self._balance -= amount
        print("[은행 시스템] 출금 완료.")
        
    def get_balance(self):
        return self._balance

# '상점'의 역할을 하는 함수
def buy_headphone(account, price):
    print(f"[상점] {price:,}원짜리 헤드폰 구매를 시도합니다.")
    
    # '상점'은 잔액이 얼마인지 모른다. 그냥 '은행'에 출금 요청만 한다.
    account.withdraw(price)
    
    print("[상점] 결제가 성공적으로 완료되었습니다!")

# --- 메인 실행 코드 ---
# 1. 내 계좌를 만든다.
my_account = BankAccount(110000)

# 2. 구매를 시도한다.
try:
    # 이 함수는 성공하거나, 실패(ValueError)하거나 둘 중 하나다.
    buy_headphone(my_account, 1000000)
except ValueError as e:
    # 은행 시스템에서 발생시킨 에러를 여기서 잡아서 처리한다.
    print(f"\n[최종 결과] 구매에 실패했습니다. 원인: {e}")

print(f"현재 잔액: {my_account.get_balance():,}원")
```
