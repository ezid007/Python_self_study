```python
import numpy as np
import matplotlib.pyplot as plt

arr = np.random.normal(0, 1, 100000)
plt.hist(arr, bins=100)
plt.title("Histogram with Mean/StdDev Lines")
plt.xlabel("Value")
plt.ylabel("Frequency")

# 평균(0) 위치에 빨간색 수직선 그리기
plt.axvline(0, color='red', linestyle='dashed', linewidth=2, label='Mean (0)')

# ±1 표준편차 위치에 주황색 수직선 그리기
plt.axvline(1, color='orange', linestyle='dashed', linewidth=2, label='+1 Std Dev')
plt.axvline(-1, color='orange', linestyle='dashed', linewidth=2)

# 범례(label) 표시
plt.legend()

plt.show()
```

```python
import numpy as np

py_list = [
    np.full(3, 8),
    np.array([33, -15, 26]),
    np.linspace(17, 26, 3)
]
```

```python
result_arr = []
for i in py_list:
    result_arr.append(np.mean(i))
    result_arr.append(np.std(i))
    result_arr.append(np.median(i))

result_arr
```

```python
import numpy as np

# 1. 2x2 기본 체스판 조각을 만듭니다.
base = np.array([[0, 1],
                [1, 0]])

# 2. 7x7을 덮을 수 있도록 4x4 크기로 이어붙입니다. (결과는 8x8)
#    그 후, 7x7 크기만큼 잘라냅니다.
result = np.tile(base, (4, 4))[:7, :7]

print(result)
```

```python
import numpy as np

# 1. 1차원 인덱스 배열을 만듭니다.
x = np.arange(7)
# 결과: [0 1 2 3 4 5 6]

# 2. 브로드캐스팅을 이용해 행 인덱스와 열 인덱스의 합을 구하고 2로 나눈 나머지를 계산합니다.
# x는 행으로, x[:, np.newaxis]는 열로 브로드캐스팅됩니다.
result = (x + x[:, np.newaxis]) % 2

print(result)
```
