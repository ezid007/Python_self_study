```python
# -*- coding: utf-8 -*-
# 필요한 라이브러리들을 불러옵니다.
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time

# with 구문을 사용하여 브라우저를 안전하게 열고 자동으로 닫습니다.
with webdriver.Chrome(service=Service(ChromeDriverManager().install())) as driver:

    # 1. 제어할 웹사이트로 이동합니다.
    driver.get("https://en.wikipedia.org/wiki/Main_Page")

    # 2. 검색창에 글자 입력하기
    search_box = driver.find_element(By.NAME, "search")
    search_box.send_keys("Python programming")
    time.sleep(1)

    # 3. XPath를 사용하여 'Search'라는 텍스트를 가진 button 찾기
    # By.XPATH: 요소를 찾기 위한 방법으로 XPath를 사용하겠다고 지정합니다.
    # "//button[text()='Search']"는 XPath 문법으로, 아래와 같은 의미입니다.
    #   //         : HTML 문서의 어디에서든 검색을 시작하라.
    #   button     : <button> 태그를 찾아라.
    #   [...]      : 그리고 다음의 추가 조건을 만족해야 한다.
    #   text()     : 그 태그가 가지고 있는 '텍스트 내용'이
    #   ='Search'  : 'Search'라는 문자열과 정확히 일치해야 한다.
    # 즉, "문서 전체에서 텍스트가 정확히 'Search'인 <button> 태그를 찾아줘" 라는 뜻입니다.
    search_button = driver.find_element(By.XPATH, "//button[text()='Search']")

    # 4. 찾은 버튼 클릭하기
    search_button.click()

    print("XPath를 이용해 텍스트로 버튼을 찾아 클릭했습니다.")
    time.sleep(3)

print("자동화 작업이 성공적으로 완료되었습니다.")
```

```python
# -*- coding: utf-8 -*-
# 필요한 라이브러리들을 불러옵니다.
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import (
    Keys,
)  # 엔터(RETURN) 같은 키보드 키를 사용하기 위해 필요합니다.
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time

# with 구문을 사용하여 브라우저를 안전하게 열고 자동으로 닫습니다.
with webdriver.Chrome(service=Service(ChromeDriverManager().install())) as driver:

    # 1. 제어할 웹사이트로 이동합니다.
    driver.get("https://en.wikipedia.org/wiki/Main_Page")

    # 2. 검색창에 글자 입력하기
    search_box = driver.find_element(By.NAME, "search")
    search_box.send_keys("Python programming", Keys.RETURN)

    # 3. 검색창에 엔터(Enter) 키를 입력하여 검색 실행하기
    # send_keys() 메서드는 글자뿐만 아니라 키보드의 특수 키도 입력할 수 있습니다.
    # 위에서 import한 Keys.RETURN은 키보드의 엔터(Enter) 키를 의미합니다.
    # 이 동작은 검색창에 텍스트를 입력한 후 엔터를 누르는 것과 정확히 같습니다.

    print("검색창에 엔터 키를 입력하여 검색을 실행했습니다.")
    time.sleep(5)

print("자동화 작업이 성공적으로 완료되었습니다.")
```

```python
# 명시적 대기를 사용한 코드
# -*- coding: utf-8 -*-
# ---------------------------------------------------------------------------
# 필요한 라이브러리들을 불러옵니다.
# ---------------------------------------------------------------------------
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

# --- 명시적 대기(Explicit Wait)를 위해 필요한 모듈 ---
# WebDriverWait: 특정 조건이 만족될 때까지 기다리는 역할을 하는 핵심 클래스입니다.
from selenium.webdriver.support.ui import WebDriverWait

# expected_conditions: "어떤 조건"을 기다릴지 정의해 둔 모음입니다. EC라는 별칭으로 사용합니다.
from selenium.webdriver.support import expected_conditions as EC

import time  # 마지막에 결과를 눈으로 확인하기 위해 사용합니다.

# ---------------------------------------------------------------------------
# 웹 드라이버 실행 및 자동화 작업
# ---------------------------------------------------------------------------

# with 구문을 사용하여 브라우저를 안전하게 열고, 작업이 끝나면 자동으로 닫습니다.
with webdriver.Chrome(service=Service(ChromeDriverManager().install())) as driver:

    # 1. 제어할 웹사이트로 이동합니다.
    driver.get("https://en.wikipedia.org/wiki/Main_Page")

    try:
        # 2. 명시적 대기를 사용하여 검색창이 나타날 때까지 기다립니다.
        # WebDriverWait(driver, 10): 드라이버를 대상으로 최대 10초까지 기다리겠다고 설정합니다.
        # .until(...): 괄호 안의 조건이 참(True)이 될 때까지 기다립니다.
        # EC.presence_of_element_located(...): 'expected_conditions' 모음 중 하나로,
        #   지정한 요소가 DOM(문서 구조)에 나타나기만 하면 기다리기를 멈추고 그 요소를 반환합니다.
        #   (By.NAME, "search"): 요소를 찾는 방법과 값입니다. 'name'이 'search'인 요소를 찾습니다.
        # 즉, "name이 'search'인 요소가 나타날 때까지 최대 10초 기다리고, 나타나면 그 요소를 search_box 변수에 넣어줘" 라는 뜻입니다.
        search_box = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.NAME, "search"))
        )

        # 3. 찾은 검색창에 검색어 입력과 엔터 키 입력을 동시에 실행합니다.
        # StaleElementReferenceException을 방지하기 위해 하나의 동작으로 처리합니다.
        search_box.send_keys("Python programming", Keys.RETURN)

        # 4. 검색 결과 페이지의 제목이 나타날 때까지 기다립니다.
        # 자동화가 성공했는지 확인하기 위한 좋은 방법입니다.
        # 'Python programming'이라는 텍스트가 페이지 제목에 포함될 때까지 최대 10초 기다립니다.
        WebDriverWait(driver, 10).until(EC.title_contains("Python programming"))

        print("명시적 대기를 사용하여 검색에 성공하고, 결과 페이지로 이동했습니다.")

        # 결과를 눈으로 확인하기 위해 잠시 대기합니다.
        time.sleep(5)

    except Exception as e:
        # 만약 10초가 지나도 요소를 찾지 못하는 등 에러가 발생하면 메시지를 출력합니다.
        print(f"자동화 도중 에러가 발생했습니다: {e}")

# with 블록이 끝나면 브라우저는 자동으로 완전히 종료됩니다.
print("자동화 작업이 성공적으로 완료되었습니다.")
```

눈에 보이는 큰 에러는 없었지만, try 블록 안의 코드 중 하나가 정해진 시간(10초) 안에 성공하지 못해서 TimeoutException이라는 '조용한' 에러가 발생했고, 그 결과 except 블록이 실행된 것입니다.


```python
# -*- coding: utf-8 -*-
# ---------------------------------------------------------------------------
# 필요한 라이브러리들을 불러옵니다.
# ---------------------------------------------------------------------------
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

# --- 명시적 대기(Explicit Wait)를 위해 필요한 모듈 ---
# WebDriverWait: 특정 조건이 만족될 때까지 기다리는 역할을 하는 핵심 클래스입니다.
from selenium.webdriver.support.ui import WebDriverWait

# expected_conditions: "어떤 조건"을 기다릴지 정의해 둔 모음입니다. EC라는 별칭으로 사용합니다.
from selenium.webdriver.support import expected_conditions as EC

import time  # 마지막에 결과를 눈으로 확인하기 위해 사용합니다.

# ---------------------------------------------------------------------------
# 웹 드라이버 실행 및 자동화 작업
# ---------------------------------------------------------------------------

# with 구문을 사용하여 브라우저를 안전하게 열고, 작업이 끝나면 자동으로 닫습니다.
with webdriver.Chrome(service=Service(ChromeDriverManager().install())) as driver:

    # 1. 제어할 웹사이트로 이동합니다.
    driver.get("https://en.wikipedia.org/wiki/Main_Page")

    try:
        # 2. 명시적 대기를 사용하여 검색창이 나타날 때까지 기다립니다.
        # name이 'search'인 요소가 DOM에 나타날 때까지 최대 10초 기다립니다.
        search_box = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.NAME, "search"))
        )

        # 3. 찾은 검색창에 검색어 입력과 엔터 키 입력을 동시에 실행합니다.
        search_box.send_keys("Python programming", Keys.RETURN)

        # 4. 검색 결과 페이지의 메인 제목(<h1>)이 나타날 때까지 기다립니다. (수정된 부분)
        # 이 방법이 페이지 제목을 확인하는 것보다 훨씬 더 안정적입니다.
        # 위키피디아의 메인 제목은 항상 id가 'firstHeading'인 h1 태그입니다.
        # EC.visibility_of_element_located: 요소가 화면에 '보일 때까지' 기다립니다.
        WebDriverWait(driver, 10).until(
            EC.visibility_of_element_located((By.ID, "firstHeading"))
        )

        print("명시적 대기를 사용하여 검색에 성공하고, 결과 페이지로 이동했습니다.")

        # 결과를 눈으로 확인하기 위해 잠시 대기합니다.
        time.sleep(5)

    except Exception as e:
        # 만약 10초가 지나도 요소를 찾지 못하는 등 에러가 발생하면 메시지를 출력합니다.
        print(f"자동화 도중 에러가 발생했습니다: {e}")

# with 블록이 끝나면 브라우저는 자동으로 완전히 종료됩니다.
print("자동화 작업이 성공적으로 완료되었습니다.")
```
