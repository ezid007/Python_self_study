```python
import requests
from datetime import datetime
```

```python
MY_LAT = 51.507351 # Your latitude
MY_LONG = -0.127758 # Your longitude

parameters = {
    "lat": MY_LAT,
    "lng": MY_LONG,
    "formatted": 0,
}
response = requests.get("https://api.sunrise-sunset.org/json", params=parameters)
response.raise_for_status()
data = response.json()
```

```python
data
```

```python
sunrise = data["results"]['sunrise'].split("T")[1].split(':')[:3]
print(f"해 뜨는 시간: {(lambda x: ':'.join(x))(sunrise)}")

```

```python
import datetime

# days=1, hours=5, minutes=30, seconds=45 로 timedelta 생성
td = datetime.timedelta(days=1, hours=24, minutes=30, seconds=45)

print(f"생성된 timedelta 객체: {td}")
print("---" * 10)

# 1. '.days'는 직접 가져올 수 있습니다.
print(f"td.days 👉 {td.days}")

# 2. '.seconds'는 hours, minutes, seconds가 합산되어 저장됩니다.
# (5시간 * 3600초) + (30분 * 60초) + 45초 = 18000 + 1800 + 45 = 19845초
print(f"td.seconds 👉 {td.seconds}")

# 3. '.hours'나 '.minutes' 속성은 존재하지 않아 오류가 발생합니다.
# print(td.hours) # AttributeError: 'datetime.timedelta' object has no attribute 'hours'

# 4. 따라서, 시간과 분은 '.seconds' 속성을 이용해 직접 계산해야 합니다.
hours = td.seconds // 3600
minutes = (td.seconds % 3600) // 60
seconds = td.seconds % 60

print("---" * 10)
print("'.seconds' 속성을 통해 계산한 시, 분, 초:")
print(f"시간: {hours}")
print(f"분: {minutes}")
print(f"초: {seconds}")
```
