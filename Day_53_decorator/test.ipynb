```python
def shout(text):
    """문자열을 대문자로 만들어 느낌표와 함께 반환하는 함수"""
    return text.upper() + "!"


def whisper(text):
    """문자열을 소문자로 만들어 반환하는 함수"""
    return text.lower()


def greet(func, message):
    """함수를 인자로 받아, 그 함수로 메시지를 처리한 후 출력하는 함수"""
    # 여기서 func는 shout 또는 whisper 함수가 됩니다.
    processed_message = func(message)
    print(processed_message)


# greet 함수에 shout 함수와 메시지를 전달
greet(shout, "Hello")

# greet 함수에 whisper 함수와 메시지를 전달
greet(whisper, "Hello")
```

```python
import time


def delay_decorator(function):
    def wrapper_function():
        time.sleep(1)
        function()
        function()

    return wrapper_function

@delay_decorator
def say_hello():
    print("Hello")

say_hello()
```
